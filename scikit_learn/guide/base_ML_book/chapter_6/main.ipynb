## **Обучение и визуализация дерева принятия решений.**

````python
from sklearn.datasets import load_iris
````

````python
from sklearn.tree import DecisionTreeClassifier
````

````python
iris = load_iris()
iris
````

````python
X = iris.data[:, 2:]  # Длина и ширина лепестка (берём все строки, но только 2 последних столбца (начиная со 2го индекса))
y = iris.target
````

````python
tree_clf = DecisionTreeClassifier(max_depth=2)
````

````python
tree_clf.fit(X, y)
````

## **Вырабатывание прогнозов.**

**Листовой узел** - узел без дочерних узлов.

Большой плюс *дерева принятия решений* - никакой предобработки данных не требуется: ни масштабирование, ни центрирование признаков.

- Атрибут `samples` узла подсчитывает, к скольким обучающим образцам он применяется.
- Атрибут `value` узла сообщает, к скольким обучающим образцам каждого класса применяется этот узел.
- Атрибут `gini` (показатель Джини) узла измеряет его **загрязнённость**: узел "чист" (`gini=0`), если все обучающие образцы, к которым он применяется, принадлежат одному и тому же классу.

### Загрязнённость Джини

$$
G_i = 1 - \sum_{k=1}^n p_{i,k}^2
$$

- $p_{i,k}$ - доля образцов класса $k$ среди обучающих образцов в $i$-том узле.

> Например, узел на глубине 2 слева имеет показатель Джини, равный $1 - (\frac{0}{54})^2 - (\frac{49}{54})^2 - (\frac{5}{54})^2 = 0.168$

`Scikit-Learn` использует алгоритм `CART`, который выпускает только двоичные деревья (нелистовые узлы имеют только два дочерних узла). Однако есть и другие алгоритмы, где дочерних узлов другое количество, например, `ID3`.

## **Оценивание вероятностей классов.**

Следует обратить внимание, что **дерево принятия решений** выдаёт одинаковую вероятность классов для всех образцов для конкретного узла.

## **Алгоритм обучения *CART*.**

**CART** - *дерево классификации и регресии*. Идея алгоритма: сначала он расщепляет обучающий набор на два поднабора с применением единственного признака $k$ и порога $t_k$, например, "длина лепестка <= 2.45 см". Алгоритм ищет пару ($k, t_k$), которая производит самые чистые поднаборы (взвешенные по их размеру)

### Функция издержек CART для классификации

$$
J(k, t_k) = \frac{m_{\text{левый}}}{m} G_{\text{левый}} + \frac{m_{\text{правый}}}{m} G_{\text{правый}},
$$

где:

- $G_{\text{левый/правый}}$ измеряет загрязнённость левого/правого поднабора,
- $m_{\text{левый/правый}}$ - количество образцов в левом/правом поднаборе.

Далее действие повторяется и алгоритм рекурсивно разбивает набор данных, пока не достигнет заданной максимальной глубины или пока не дойдёт до момента, когда уже эффективно расщепить данные не получится.

CART является "жадным" алгоритмом - он не анализирует эффективность расщепления дальше, чем один уровень. *Нахождение оптимального дерева* - *NP-полная* задача, которая требует *полиномиального* решения (`O(exp(m))`).

## **Вычислительная сложность.**

Если алгоритм проходится по дереву и смотрит на 1 признак, то его сложность очень даже хорошая - `O(log2(m))`. Однако алгоритм сравнивает все признаки (если `max_features` по умолчанию) на всех образцах на каждом узле, а это уже - `O(x * m * log2(m))`.

На небольшом наборе можно ускорить обучение за счёт предварительной сортировки данных (`presort=True`), но это значительно замедлит обучение при более крупных обучающих наборах.

## **Загрязнённость Джини или энтропия?**

Установив `criterion="entropy"`, можно сменить меру неопределённости с `gini` на **этропию**. Концепция энтропии появилась в термодинамике как мера молекулярного беспорядка: энтропия приближается к нулю, когда молекулы неподвижны и вполне упорядочены. Энтропия набора равна нулю, когда он содержит образцы только одного класса.

### Энтропия $i$-того узла

$$
H_i = - \sum_{k=1}^n p_{i,k}log(p_{i,k})
$$

$$
p_{i,k} \neq 0
$$

> Например, узел на глубине 2 слева имеет энтропию равную $- \frac{49}{54}log(\frac{49}{54}) - \frac{5}{54}log(\frac{5}{54}) = 0.31$

Загрязненность Джини слегка быстрее подсчитывать, однако она имеет тенденцию изолировать самый часто встречающийся класс в собственной ветви дерева, а энтропия - производить чуть более сбалансированные деревья.

## **Гиперпараметры регуляризации.**

- *Непараметрическая модель* - модель, количество параметров которой перед обучением не определено, оттого структура модели вольна тесно привязываться к данным.
- *Параметрическая модель* - (подобно линейной модели) имеет предопределённое количество количество параметров, так что её степень свободы ограничивается, сокращая риск переобучения (но увеличивая риск недообучения).

Параметры `DecisionTreeClassifier` влияющие на степень свободы модели:
- `min_samples_split` - минимальное количество образцов, которые должны присутствовать в узле, прежде чем его можно будет расщепить;
- `min_samples_leaf` - минимальное количество образцов, которое должно иметь листовой узел;
- `min_weight_fraction_leaf` - аналог `min_samples_leaf`, выраженный в виде доли от общего числа взвешенных образцов;
- `max_leaf_nodes` - максимальное количество листовых узлов;
- `max_features` - максимальное число признаков, которые оцениваются при расщеплении каждого узла.

Увеличение гиперпараметров `min_*` или уменьшение гиперпараметров `max_*` будет регуляризировать модель.

Другие алгоритмы работают, сначала обучая дерево принятия решений без ограничений и затем *отсекая* (удаляя) излишние узлы. Узел, все дочерние узлы которого листовые, считается излишним, если обеспечиваемое им улучшение чистоты не является *статистически значимым*. Стандартные статистические тесты, такие как *тест $X^2$*, применяются для оценкивероятности того, что улучшение представляет собой исключительно результат случайности (которая называется *нулевой гипотезой*). Если эта вероятность, называемая *$р$-значением*, выше заданного порога (обычно составляющего 5% и управляемого гиперпараметром), тогда узел считается излишним и его дочерние узлы удаляются. Отсечение продолжается до тех пор, пока не будут удалены все излишние узлы.

Следует обратить внимание на то, что спрогнозированное значение для каждой области всегда будет средним целевым значением образцов в этой области.

Алгоритм CART работает главным образом так же, как раньше, но только вместо попытки расщеплять обучающий набор способом, сводящим к минимуму загрязненность, он пробует расщеплять его способом, который минимизирует MSE.

### Функция издержек алгоритма CART для регрессии

$$
J(k, t_k) = \frac{m_{\text{левый}}}{m} MSE{\text{левый}} + \frac{m_{\text{правый}}}{m} MSE{\text{правый}},
$$

где:

- $MSE_{\text{узел}} = \sum_{i \in \text{узел}} (\hat{y}_{\text{узел}} - y^{(i)})^2$
- $\hat{y}_{\text{узел}} = \frac{1}{m_{\text{узел}}} \sum_{i \in \text{узел}} y^{(i)}$

## **Неустойчивость.**

Ограничения:

- Прежде всего, деревья принятия решений предпочитают прямоугольные границы решений (все расщепления перпендикулярны той или иной оси), которые делают их чувствительными к поворотам обучающего набора

- Деревья принятия решений сильно чувствительны к небольшим изменениям в обучающих данных. В действительности, поскольку применяемый библиотекой `Scikit-Learn` алгоритм обучения является *стохастическим*, можно получать очень разные модели даже на тех же самых обучающих данных (если только не установлен `random_state`). **Случайные леса** могут ограничить эту неустойчивость путем *усреднения прогнозов по многим деревьям*.
