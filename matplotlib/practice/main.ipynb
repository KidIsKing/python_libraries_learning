## 🧭 Сводная таблица: какой тип графика когда использовать

| Тип графика | Метод | Когда применять |
|------------|-------|----------------|
| Линейный | `plt.plot()` | Функции, временные ряды, тренды |
| Гистограмма | `plt.hist()` | Распределение одной переменной |
| Круговая | `plt.pie()` | Доли категорий (≤5-7 категорий) |
| Столбчатая | `plt.bar()` | Сравнение категорий по значению |
| Рассеяния | `plt.scatter()` | Связь двух переменных, кластеры |
| Плотность | `plt.hexbin()` | Плотность точек при больших данных |
| Область между | `plt.fill_between()` | Интервалы, неопределённость, разница функций |
| Полярный | `projection='polar'` | Угловые данные, периодические процессы |

```python
import matplotlib.pyplot as plt
import numpy as np
```

## 📌 Задача A: Линейный график `y = 2x + 1`

### 📋 Техническое задание
- Построить график функции `y = 2x + 1` для `x ∈ [-10, 10]` с шагом `0.2`
- Добавить: заголовок, подписи осей, легенду, сетку

```python
x = np.arange(-10, 10.2, 0.2)
y = 2 * x + 1

plt.plot(x, y, label="$y = 2 * x + 1$")
plt.grid()
plt.legend()

plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("$y = 2 * x + 1$")
```

## 📌 Задача B: Гистограмма нормального распределения

### 📋 Техническое задание
- Сгенерировать 50 чисел из `N(μ=0, σ=1)`
- Построить гистограмму: `bins=10`, `alpha=0.5`, `color='blue'`

```python
mean = 0  # Среднее значение
std_dev = 1  # Стандартное отклонение
num_samples = 50

# normal_data = np.random.normal(loc=0, scale=1, size=50)
normal_data = np.random.normal(mean, std_dev, num_samples)

plt.hist(normal_data, bins=10, alpha=0.5, color="blue")
```

## 📌 Задача C: Круговая диаграмма

### 📋 Техническое задание
- Данные: `['A','B','C','D']`, значения `[25,35,15,25]`
- Выделить сектор 'B', показать проценты, добавить заголовок

```python
values = [25, 35, 15, 25]
labels = ["A", "B", "C", "D"]
explode = [0, 0.1, 0, 0]  # Сдвиг для "B"

plt.pie(values, labels=labels, explode=explode, autopct="%1.1f%%")

plt.title("Круговая диаграмма")
```

## 📌 Задача D: Столбчатая диаграмма

### 📋 Техническое задание
- Категории: `['Яблоки','Бананы','Вишня','Финики']`, значения `[10,20,15,5]`
- Индивидуальные цвета столбцов, подписи осей, заголовок

```python
categories = ["Яблоки", "Бананы", "Вишня", "Финики"]
values = [10, 20, 15, 5]

plt.bar(categories, values, color=["red", "yellow", "purple", "black"])

plt.title("Столбчатая диаграмма")
```

## 📌 Задача E: Диаграмма рассеяния с цветовой картой

### 📋 Техническое задание
- 100 случайных точек `(x,y) ∈ [0,50]`
- Размер точки ∝ `y`, цвет по карте `viridis` в зависимости от `y`
- Добавить colorbar, подписи, заголовок

```python
np.random.seed(42)  # Для сохранения генерации точек в одних и тех же местах
x = np.random.uniform(0, 50, 100)
y = np.random.uniform(0, 50, 100)
sizes = y * 10

scatter = plt.scatter(x, y, s=sizes, c=y, cmap='viridis')

plt.xlabel('Значение x')
plt.ylabel('Значение y')
plt.title('Диаграмма рассеяния с цветовой картой')
plt.colorbar(scatter, label='Значение y')
```

## 📌 Задача F: Два графика на одной системе координат

### 📋 Техническое задание
- Построить `y₁ = x²` и `y₂ = -x² + 100` для `x ∈ [-10, 10]`
- Разные стили линий, легенда, сетка, заголовок

### Пояснение к стилям:

1. `'-'` - сплошная
2. `'--'` - пунктир
3. `'-.'` - штрихпунктир

```python
x = np.arange(-10, 10, 0.1)
y1 = x ** 2
y2 = -x ** 2 + 100

plt.plot(x, y1, label="$y_1 = x^2$", linestyle="-")
plt.plot(x, y2, label="$y_2 = -x^2 + 100$", linestyle="--")
plt.grid()
plt.legend()
plt.title("Два графика вместе")
```

## 📌 Задача G: Диаграмма плотности (hexbin)

### 📋 Техническое задание
- 500 точек из `N(0,1)` для `x` и `y`
- Hexbin-диаграмма с картой `plasma`, colorbar, подписи

```python
x = np.random.normal(0, 1, 500)
y = np.random.normal(0, 1, 500)

plt.hexbin(x, y, cmap='plasma', gridsize=30)
plt.colorbar(label='Плотность')
plt.title('Диаграмма плотности точек')
plt.xlabel('Значение x')
plt.ylabel('Значение y')
```

## 📌 Задача H: Синус и косинус (исправленная версия)

### 📋 Техническое задание
- `y = sin(x)` и `y = cos(x)` для `x ∈ [0, 2π]`
- Синус: синяя сплошная, косинус: красная штриховая

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, color="blue", linestyle="-")
plt.plot(x, y2, color="red", linestyle="--")
```

## 📌 Задача I: Область между графиками

### 📋 Техническое задание
- `y₁ = x²`, `y₂ = x³` для `x ∈ [-2, 2]`
- Заштриховать область между ними: `color='red'`, `alpha=0.2`

```python
x = np.linspace(-2, 2, 50)
y1 = x ** 2
y2 = x ** 3

plt.plot(x, y1)
plt.plot(x, y2)
plt.fill_between(x, y1, y2, color="red", alpha=0.2)
```

## 📌 Задача J: Полярный график (кардиоида)

### 📋 Техническое задание
- Полярная функция: `r = 1 + cos(θ)`, `θ ∈ [0, 2π]`
- Синяя линия толщиной 2, заголовок с символом θ

```python
theta = np.linspace(0, 2*np.pi + 1, 100)
r = 1 + np.cos(theta)

plt.figure(figsize=(8, 8))

ax = plt.subplot(111, projection='polar')
ax.plot(theta, r, color='blue', linewidth=2)
ax.set_title('Полярный график функции r = 1 + cos(θ)', pad=20)
```
