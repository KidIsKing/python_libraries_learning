## **Основы**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [22, 5, 89, -10, 1]

plt.plot(x, y)
plt.xlabel("Ox")
plt.ylabel("Oy")
plt.title("Func")
plt.grid()  # Сетка
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

# Можно самим устанавливать цвета
plt.plot(x, y1, label="y1", color="red")
plt.plot(x, y2, label="y2", color="#bebebe")
plt.legend()
```

```python
# Более реальное применение
x = [i / 20 for i in range(100)]
y = [i ** 2 for i in x]

plt.plot(x, y)
```

```python
# С NumPy
x = np.linspace(0, 5, 100)
y = x ** 0.5
plt.plot(x, y)
```

```python
# Всё вместе
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [2, 4, 6, 8, 10]

plt.plot(x, y1, label="Project 1", color="red")
plt.plot(x, y2, label="Project 2", color="#bebebe")
plt.legend()

plt.xlabel("Days")
plt.ylabel("NNet Profit ($M)")
plt.title("Stonks")
plt.grid()
```

## **Диаграммы**

```python
y = np.random.randint(1, 100, 100)

# Построение гистограммы
plt.hist(y, bins=30)  # По умолчанию bins=10 (количество столбцов)
```

```python
y = np.random.randint(1, 100, 100)

# Ящик с усами
plt.boxplot(y)
```

```python
# Если данные дискретные, их можно показать на столбчатой диаграмме (барчарт)

x = ["Windows", "macOS", "Linux", "Other"]
h = [68, 14, 3, 15]  # Высота

plt.bar(x, h)
```

```python
# Некоторые данные лучше смотрятся на горизонтальных столбцах

x = ["Windows", "macOS", "Linux", "Other"]
h = [68, 14, 3, 15]  # Высота

plt.barh(x[::-1], h[::-1])  #  Так как столбцы отрисуются по возрастанию, резонно развернуть списки
```

```python
x = ["Windows", "macOS", "Linux", "Other"]
h = [68, 14, 3, 15]  # Высота

# Круговая диаграмма
plt.pie(h, labels=x)
```

```python
x = [1, 10, 100, 1000]
y = [1, 2, 3, 4]
plt.semilogx(x, y)  # Логарифмический масштаб по оси X
```

```python
x = [1, 2, 3, 4]
y = [1000, 100, 10, 1]
plt.semilogy(x, y)  # Логарифмический масштаб по оси Y
```

```python
x = [1, 10, 100, 1000]
y = [1000, 100, 10, 1]
plt.loglog(x, y)  # Логарифмический масштаб по обоим осям
```

```python
x = np.linspace(0.2, 2)
y1 = x ** 2
y2 = 1 / x

plt.plot(x, y1)
plt.plot(x, y2)

# Маркер чёрного цвета (по умолчанию - синего)
plt.axvline(1, color="black")  # Вертикальный
plt.axhline(1, color="black")  # Горизонтальный
```

## **Кастомизация**

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

plt.plot(x, y1, linewidth=0.5)
plt.plot(x, y2)  # По умолчанию ширина 1
plt.plot(x, y3, linewidth=2)
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

plt.plot(x, y1, linestyle="dashed")  # Пунктир
plt.plot(x, y2, linestyle="dotted")  # Точка
plt.plot(x, y3, linestyle="dashdot")  # Пунктир с точкой
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

# Каждую точку на графике можно выделить маркерами разной формы
plt.plot(x, y1, marker="o")  # Круг
plt.plot(x, y2, marker="v")  # Треугольник
plt.plot(x, y3, marker="d")  # Ромб
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

# Каждую точку на графике можно выделить маркерами разной формы
plt.plot(x, y1, marker="o", markersize=3)
plt.plot(x, y2, marker="v", markersize=10)
plt.plot(x, y3, marker="d")  # По умолчанию размер маркера равен 5
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

# Для графика из точек также можно изменить вид маркеров и размер (но аргумент уже "s")
plt.scatter(x, y1, marker="o", s=3)
plt.scatter(x, y2, marker="v", s=10)
plt.scatter(x, y3, marker="d")  # По умолчанию размер маркера равен 20
```

```python
x = [1, 2, 3]
y = [1, 2, 3]

plt.scatter(x, y, c="red") # Можно выбрать цвет
```

```python
x = [1, 2, 3]
y = [1, 2, 3]

plt.scatter(x, y, c=["red", "yellow", "green"]) # Можно выбрать цвет для каждой точки
```
